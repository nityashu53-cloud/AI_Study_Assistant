```python
def study_assistant(question):
    if not question.strip():
        return "Please enter a question."

    prompt = f"""
You are an AI Study Assistant for college students.

Answer the student's question in a clear and simple way.
Use examples where helpful.
Avoid unnecessary complicated language.

Student's Question:
{question}
"""

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )

    return response.text
```

```python
question = input("Enter your study question: ")

answer = study_assistant(question)

print("\nAI Study Assistant:")
print(answer)
```

```python
def study_assistant(mode, user_input):

    if not user_input.strip():
        return "Please enter some text."

    if mode == "Explain Topic":
        prompt = f"""
        Explain the following topic to a college student in simple language.
        Give a clear explanation and one simple example.

        Topic:
        {user_input}
        """

    elif mode == "Summarize Notes":
        prompt = f"""
        Summarize the following study notes.
        Give the important points in simple bullet points.

        Notes:
        {user_input}
        """

    elif mode == "Generate Quiz":
        prompt = f"""
        Create 5 multiple-choice questions based on the following topic.
        Give 4 options for each question and clearly mention the correct answer.

        Topic:
        {user_input}
        """

    elif mode == "Ask Question":
        prompt = f"""
        Answer the following student's question clearly and simply.
        Give examples where helpful.

        Question:
        {user_input}
        """

    else:
        return "Please select a valid option."

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=prompt
    )

    return response.text
```

```python
print("===== AI STUDY ASSISTANT =====")
print("1. Explain Topic")
print("2. Summarize Notes")
print("3. Generate Quiz")
print("4. Ask Question")

choice = input("\nEnter your choice (1-4): ")

options = {
    "1": "Explain Topic",
    "2": "Summarize Notes",
    "3": "Generate Quiz",
    "4": "Ask Question"
}

if choice not in options:
    print("Invalid choice. Please select 1, 2, 3, or 4.")
else:
    mode = options[choice]
    user_input = input(f"\nEnter your {mode.lower()} input:\n")

    answer = study_assistant(mode, user_input)

    print("\n===== AI RESPONSE =====")
    print(answer)
```
